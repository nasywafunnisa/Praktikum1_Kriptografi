```python
alphabet = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
substitusi = "QWERTYUIOPASDFGHJKLZXCVBNM"

def encrypt(text):
    return "".join(substitusi[alphabet.index(ch)] if ch in alphabet else ch
                  for ch in text.upper())

def decrypt(text):
    return "".join(alphabet[substitusi.index(ch)] if ch in substitusi else ch
                  for ch in text.upper())

plain = "KRIPTOGRAFI"
cipher = encrypt(plain)
back = decrypt(cipher)

print("plaintext : ", plain)
print("Ciphertext : ", cipher)
print("Deskripsi : ", back)
```
