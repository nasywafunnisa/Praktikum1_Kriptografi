```python
import base64

def cek_bit(bits, panjang):
    return len(bits) == panjang and all(b in "01" for b in bits)

def xor_bits(a, b):
    return ''.join(
        '1' if x != y else '0'
        for x, y in zip(a, b)
    )

def f_function(R, K):
    return xor_bits(R, K)

def feistel_round(L, R, K):
    F = f_function(R, K)
    new_L = R
    new_R = xor_bits(L, F)
    return new_L, new_R

def feistel_encrypt_byte(byte_value, keys):
    bits = format(byte_value, '08b')
    L = bits[:4]
    R = bits[4:]

    for K in keys:
        L, R = feistel_round(L, R, K)

    encrypted_bits = L + R
    return int(encrypted_bits, 2)

def encrypt_message(message, keys):
    data = message.encode('utf-8')
    encrypted = bytes(
        feistel_encrypt_byte(b, keys)
        for b in data
    )
    ciphertext = base64.b64encode(encrypted).decode('ascii')
    return ciphertext

print('=====================================')
print(' FEISTEL NETWORK - PESAN TEKS')
print('=====================================')
message = input('Masukkan pesan: ')
K1 = input('Masukkan K1 (4 bit): ')
K2 = input('Masukkan K2 (4 bit): ')
K3 = input('Masukkan K3 (4 bit): ')

keys = [K1, K2, K3]
if not all(cek_bit(k, 4) for k in keys):
    print('\nError: setiap key harus terdiri dari 4 bit (0 dan 1).')
    raise SystemExit
ciphertext = encrypt_message(message, keys)

print('\n=====================================')
print('HASIL ENKRIPSI')
print('=====================================')
print('Pesan asli  :', message)
print('K1          :', K1)
print('K2          :', K2)
print('K3          :', K3)
print('Ciphertext  :', ciphertext)
```

```python
import base64

SBOX = {
    "0000":"1110", "0001":"0100", "0010":"1101", "0011":"0001",
    "0100":"0010", "0101":"1111", "0110":"1011", "0111":"1000",
    "1000":"0011", "1001":"1010", "1010":"0110", "1011":"1100",
    "1100":"0101", "1101":"1001", "1110":"0000", "1111":"0111"
}
INV_SBOX = {v: k for k, v in SBOX.items()}
PBOX = [1, 5, 9, 13, 2, 6, 10, 14,
        3, 7, 11, 15, 4, 8, 12, 16]

def cek_bit(bits, panjang):
    return len(bits) == panjang and all(b in "01" for b in bits)

def xor_bits(a, b):
    return ''.join('1' if x != y else '0'
                   for x, y in zip(a, b))

def sbox_transform(state):
    hasil = ''
    for i in range(0, len(state), 4):
        nibble = state[i:i+4]
        hasil += SBOX[nibble]
    return hasil

def inverse_sbox_transform(state):
    hasil = ''
    for i in range(0, len(state), 4):
        nibble = state[i:i+4]
        hasil += INV_SBOX[nibble]
    return hasil

def pbox_transform(state):
    return ''.join(state[pos - 1] for pos in PBOX)

def inverse_pbox_transform(state):
    hasil = [''] * 16
    for j, pos_input in enumerate(PBOX):
        hasil[pos_input - 1] = state[j]
    return ''.join(hasil)

def spn_round(state, key):
    state = sbox_transform(state)
    state = pbox_transform(state)
    state = xor_bits(state, key)
    return state

def spn_round_inverse(state, key):
    state = xor_bits(state, key)
    state = inverse_pbox_transform(state)
    state = inverse_sbox_transform(state)
    return state

def pad_data(data):
    pad_len = 2 - (len(data) % 2)
    return data + bytes([pad_len]) * pad_len

def unpad_data(data):
    pad_len = data[-1]
    if pad_len < 1 or pad_len > 2:
        raise ValueError('Padding tidak valid.')
    return data[:-pad_len]

def encrypt_block(block, K0, K1, K2):
    state = format(int.from_bytes(block, 'big'), '016b')
    state = xor_bits(state, K0)
    state = spn_round(state, K1)
    state = spn_round(state, K2)
    return int(state, 2).to_bytes(2, 'big')

def decrypt_block(block, K0, K1, K2):
    state = format(int.from_bytes(block, 'big'), '016b')
    state = spn_round_inverse(state, K2)
    state = spn_round_inverse(state, K1)
    state = xor_bits(state, K0)
    return int(state, 2).to_bytes(2, 'big')

def encrypt_message(message, K0, K1, K2):
    data = pad_data(message.encode('utf-8'))
    hasil = b''
    for i in range(0, len(data), 2):
        block = data[i:i+2]
        hasil += encrypt_block(block, K0, K1, K2)
    return base64.b64encode(hasil).decode('ascii')

def decrypt_message(ciphertext_text, K0, K1, K2):
    encrypted = base64.b64decode(ciphertext_text.encode('ascii'))
    if len(encrypted) % 2 != 0:
        raise ValueError('Panjang ciphertext tidak sesuai dengan blok 16 bit.')
    hasil = b''
    for i in range(0, len(encrypted), 2):
        block = encrypted[i:i+2]
        hasil += decrypt_block(block, K0, K1, K2)
    hasil = unpad_data(hasil)
    return hasil.decode('utf-8')

print('=== TOY SPN 16-BIT - PESAN TEKS ===')
message = input('Masukkan pesan: ')
K0 = input('Masukkan K0 (16 bit): ')
K1 = input('Masukkan K1 (16 bit): ')
K2 = input('Masukkan K2 (16 bit): ')

if not (cek_bit(K0, 16) and cek_bit(K1, 16) and cek_bit(K2, 16)):
    print('Error: K0, K1, dan K2 harus terdiri dari 16 bit (0 dan 1).')
    raise SystemExit

ciphertext = encrypt_message(message, K0, K1, K2)
plaintext_kembali = decrypt_message(ciphertext, K0, K1, K2)

print('\nPesan asli       :', message)
print('Ciphertext teks  :', ciphertext)
print('Hasil dekripsi   :', plaintext_kembali)
print('Verifikasi       :', plaintext_kembali == message)
```

```python
def valid_bits(bits, length):
    return len(bits) == length and all(b in "01" for b in bits)

def input_bits(prompt, length):
    while True:
        bits = input(prompt).strip()
        if valid_bits(bits, length):
            return bits
        print(f"Input harus terdiri dari {length} bit (hanya 0 dan 1).")

def xor_bits(a, b):
    return ''.join('0' if x == y else '1' for x, y in zip(a, b))

def f_function(R, K):
    return xor_bits(R, K)

def feistel_table(L, R, keys):
    tabel = []
    for i, K in enumerate(keys, start=1):
        F = f_function(R, K)
        new_L = R
        new_R = xor_bits(L, F)
        tabel.append((i, L, R, F, K, new_L, new_R))
        L, R = new_L, new_R
    return L, R, tabel

L = input_bits("Masukkan L0 (4 bit): ", 4)
R = input_bits("Masukkan R0 (4 bit): ", 4)
keys = [input_bits(f"Masukkan K{i} (4 bit): ", 4) for i in range(1, 5)]

L, R, tabel = feistel_table(L, R, keys)

print("\n" + "-" * 62)
print(f"{'Round':<7}{'L lama':<9}{'R lama':<9}{'F':<7}{'Key':<7}{'L baru':<9}{'R baru':<9}")
print("-" * 62)
for i, l, r, f, k, nl, nr in tabel:
    print(f"{i:<7}{l:<9}{r:<9}{f:<7}{k:<7}{nl:<9}{nr:<9}")
print("-" * 62)
print("Ciphertext =", L + R)
```

```python
def valid_bits(bits, length):
    return len(bits) == length and all(b in "01" for b in bits)

def input_bits(prompt, length):
    while True:
        bits = input(prompt).strip()
        if valid_bits(bits, length):
            return bits
        print(f"Input harus terdiri dari {length} bit (hanya 0 dan 1).")

def xor_bits(a, b):
    return ''.join('0' if x == y else '1' for x, y in zip(a, b))

def f_function(R, K):
    return xor_bits(R, K)

def feistel_encrypt(block, keys):
    L, R = block[:4], block[4:]
    print(f"Awal : L = {L}, R = {R}")
    for i, K in enumerate(keys, start=1):
        F = f_function(R, K)
        L, R = R, xor_bits(L, F)
        print(f"Round {i}: L = {L}, R = {R}, F = {F}, K = {K}")
    return L + R

def feistel_decrypt(block, keys):
    # Tukar posisi L dan R pada ciphertext, gunakan key terbalik, lalu tukar kembali
    L, R = block[4:], block[:4]
    print(f"Awal : L = {L}, R = {R}")
    for i, K in enumerate(reversed(keys), start=1):
        F = f_function(R, K)
        L, R = R, xor_bits(L, F)
        print(f"Round {i}: L = {L}, R = {R}, F = {F}, K = {K}")
    return R + L

plaintext = input_bits("Masukkan plaintext (8 bit): ", 8)
keys = [input_bits(f"Masukkan K{i} (4 bit)    : ", 4) for i in range(1, 4)]

print("\n=== ENKRIPSI ===")
ciphertext = feistel_encrypt(plaintext, keys)
print("Ciphertext =", ciphertext)

print("\n=== DEKRIPSI (key terbalik: K3, K2, K1) ===")
hasil = feistel_decrypt(ciphertext, keys)
print("Hasil dekripsi =", hasil)

print("\n=== PEMBUKTIAN ===")
print("Plaintext awal  =", plaintext)
print("Hasil dekripsi  =", hasil)
print("Sama?           =", plaintext == hasil)
```
